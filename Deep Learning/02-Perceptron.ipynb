## What is a Perceptron?

**Perceptron is nothing but a algorithm which is used in supervised machine learning.**

A Perceptron is the simplest artificial neural network model used for binary classification.

It takes multiple input features, multiplies them by weights, adds a bias, and passes the result through an activation function to produce an output (0 or 1).

- x → input features

- w → weights

- b → bias

- f → activation function (usually step function)

- y → output

## Perceptron vs Biological Neuron

| Biological Neuron                  | Perceptron                          |
| ---------------------------------- | ----------------------------------- |
| Dendrites receive signals          | Inputs receive values               |
| Synapses control signal strength   | Weights control importance          |
| Cell body sums signals             | Weighted sum is calculated          |
| Axon fires signal                  | Activation function produces output |
| Works with electrochemical signals | Works with numerical values         |


Key Difference

- Neuron → biological and complex

- Perceptron → mathematical and simplified model

**A perceptron is an inspired abstraction, not an exact copy.**

## What is the Perceptron Trick?

The Perceptron Trick converts the classification problem into a geometric problem by expressing the decision boundary as a dot product between weight vector and input vector.


	​


### Why This Trick is Powerful (Intuition)

- Removes special handling of bias

- Simplifies learning rule

- Makes math cleaner

- Helps visualize classification as vector alignment

## Perceptron Loss Function

The Perceptron Loss Function measures how wrong a prediction is when a data point is misclassified.
Unlike MSE or Cross-Entropy, it penalizes only incorrect predictions.
