# What is NLP Pipeline?

Natural Language Processing (NLP) pipeline is a **sequence of steps** used to build end-to-end NLP software.
It outlines the complete workflow from collecting raw text to deploying a machine learning model.



---

## 🔹 Steps in NLP Pipeline

### **1. Data Acquisition**

Collecting raw text data from various sources such as websites, documents, APIs, social media, etc.

---

### **2. Text Preparation**

Transforming raw text into clean, structured, and usable form.

#### ✔ Text Cleanup

* Removing noise (HTML tags, URLs, punctuation, numbers, emojis)
* Converting text to lowercase
* Removing stopwords

#### ✔ Basic Preprocessing

* Tokenization
* Lemmatization/Stemming
* Spelling correction
* Normalization

#### ✔ Advanced Preprocessing

* Handling negations
* Handling slang/short forms
* Named Entity Correction
* Context-based normalization

---

### **3. Feature Engineering**

Converting processed text into numerical features for model training.

Examples:

* Bag of Words (BoW)
* TF-IDF
* Word Embeddings (Word2Vec, GloVe)
* Sentence Embeddings (BERT)

---

### **4. Modelling**

Building machine learning or deep learning models on numerical text features.

#### ✔ Model Building

* Selecting algorithms (Naive Bayes, SVM, LSTM, Transformers)
* Training the model

#### ✔ Evaluation

* Measuring model performance
* Metrics: Accuracy, Precision, Recall, F1-score, ROC-AUC, etc.

---

### **5. Deployment**

Making the trained NLP model available for real-world use.

#### ✔ Deployment

* Flask/FastAPI
* Docker
* Cloud deployment

#### ✔ Monitoring

* Tracking model performance
* Detecting model drift

#### ✔ Model Update

* Re-training with new data
* Improving accuracy and relevance
