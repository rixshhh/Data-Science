# **Natural Language Processing (NLP) – Notes**

## **What is NLP?**

Natural Language Processing (NLP) is a subfield of:

* **Linguistics**
* **Computer Science**
* **Artificial Intelligence**

NLP focuses on the **interaction between computers and human language**.
Its goal is to **enable computers to understand, process, and analyze large amounts of natural language data**.

---

## **Real-World Applications of NLP**

### **1. Contextual Advertisements**

* Targeted ads based on user search history, content, or preferences.

### **2. Email Clients**

* **Spam filtering**
* **Smart reply generation**
* **Auto-categorization**

### **3. Social Media**

* Detecting and removing adult/abusive content
* Opinion mining & sentiment analysis
* Fake news detection

### **4. Search Engines**

* Query understanding
* Ranking relevant results
* Auto-completion

### **5. Chatbots & Virtual Assistants**

* Customer support automation
* Voice-based assistants like Siri, Alexa
* Helpdesk bots

---

## **Approaches to NLP**

### **1. Heuristic Methods (Rule-Based NLP)**

* Uses manually created linguistic rules
* Examples: Grammar checkers, stemming using rules

### **2. Machine Learning Based Methods**

* Uses statistical or classical ML algorithms
* Examples:

  * Naive Bayes
  * SVM
  * Logistic Regression
  * Hidden Markov Models

### **3. Deep Learning Based Methods**

* Uses neural networks
* Examples:

  * RNNs, LSTMs, GRUs
  * Transformers (BERT, GPT, T5)
  * Sequence-to-sequence models

---

##  **Common NLP Tasks**

### **1. Text Classification**

* Sentiment analysis
* Spam detection
* Topic labeling

### **2. Tokenization**

* Splitting text into sentences or words

### **3. Named Entity Recognition (NER)**

* Extracting names, dates, locations, organizations

### **4. Part-of-Speech (POS) Tagging**

* Identifying nouns, verbs, adjectives, etc.

### **5. Machine Translation**

* Translating text from one language to another

### **6. Text Summarization**

* Extractive (select key sentences)
* Abstractive (generate a summary)

### **7. Question Answering**

* Reading comprehension models
* Chatbots

### **8. Text Generation**

* Autocomplete, sentence generation
* Large Language Models (LLMs)

### **9. Information Retrieval**

* Search engines
* Document ranking

### **10. Speech Recognition & Processing**

* Converting speech to text
* Voice assistants

---

## **Challenges in NLP**

### **1. Ambiguity**

* Words or sentences with multiple meanings
  *Example:*
  *I saw the boy on the beach with my binoculars.*

### **2. Contextual Words**

* Meaning depends on context
  *Example:*
  *I ran to the store because we ran out of milk.*

### **3. Colloquialisms & Slang**

* Informal expressions
  *Example:* “piece of cake”, “pulling your leg”

### **4. Synonyms**

* Multiple words with the same meaning
  *Example:* big, huge, large

### **5. Irony, Sarcasm & Tone**

* Hard for computers to detect
  *Example:* *That’s just what I needed today!*

### **6. Spelling Errors**

* Typographical mistakes
  *Example:* “enviroment” for “environment”

### **7. Creativity**

* Hard to interpret poems, stories, dialogues

### **8. Language Diversity**

* Dialects, multilingual text, code-mixed language
