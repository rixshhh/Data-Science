### 1. Word & Sentence Tokenization using NLTK

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

# Download pretrained models (for sentence tokenizer)
nltk.download('punkt')

text = "Hello everyone. Welcome to this Repositary! NLP is fun, isn't it?"

# Sentence tokenization
sentences = sent_tokenize(text)
print("Sentences:", sentences)


# Word tokenization
words = word_tokenize(text)
print("Word Tokens:", words) 
```

Description:

* `sent_tokenize` splits text into a list of sentences.

* `word_tokenize` splits a sentence into words (and punctuation as separate tokens).

This is the most common approach for languages like English with clear word boundaries

---

### 2. Tokenization using spaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Apple is looking at buying U.K. startup for $1 billion."

doc = nlp(text)

# Word tokens
tokens = [token.text for token in doc]
print("spaCy tokens:", tokens)
# e.g. ['Apple', 'is', 'looking', 'at', 'buying', 'U.K.', 'startup', 'for', '$', '1', 'billion', '.']

# Sentence segmentation
sentences = [sent.text for sent in doc.sents]
print("spaCy sentences:", sentences)
# ['Apple is looking at buying U.K. startup for $1 billion.']
```

Description:

* spaCy‘s tokenizer is rule-based plus heuristics: faster and often more robust (handles punctuation, abbreviations, special tokens).

* Good for preprocessing before tasks like NER / POS tagging / parsing.

### 3. Sub-word Tokenization using HuggingFace Tokenizers (e.g. BPE / WordPiece)

Modern NLP / Deep-Learning pipelines often use sub-word tokenization to handle rare words and to keep vocabulary manageable.

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.trainers import BpeTrainer

# Initialize a Byte-Pair Encoding tokenizer
tokenizer = Tokenizer(BPE())
tokenizer.pre_tokenizer = Whitespace()

# Sample training data (in real use-case: large corpus)
corpus = [
    "tokenization is crucial for nlp",
    "handling out-of-vocabulary words is easier with subword tokenization",
    "subword units help reduce vocabulary size"
]

trainer = BpeTrainer(special_tokens=["<unk>"])
tokenizer.train_from_iterator(corpus, trainer)

# Use the tokenizer
encoded = tokenizer.encode("subword tokenization helps in handling rarewords")
print("Sub-word tokens:", encoded.tokens)
# Example output: ['sub', 'word', 'Ġtokenization', 'Ġhelps', 'Ġin', 'Ġhandling', 'Ġrare', 'words']
```

Why use sub-words?

* Handles rare / unseen words without exploding vocabulary size.

* Helps generalize across word formations (e.g. “running”, “runner”, “runs”).

### 4. Character-level Tokenization (Custom / Simple)

```python
text = "Tokenization"

char_tokens = list(text)
print("Character Tokens:", char_tokens)
# ➤ ['T', 'o', 'k', 'e', 'n', 'i', 'z', 'a', 't', 'i', 'o', 'n']
```

When is this useful?

* For tasks like spelling correction, OCR error correction, character-level language models, or languages with no clear word boundaries.

---

### 5. Regex-based / Rule-based Tokenization (custom logic)

Sometimes you might want full control: remove punctuation, ignore certain tokens, treat special sequences as one token, etc.

```python
import re

text = "Hello!!! Welcome to NLP 101. Let's tokenize: email@example.com, URLs like https://example.com"

# Basic regex to extract words (letters + digits)
tokens = re.findall(r"\b\w+\b", text)
print("Regex tokens:", tokens)
# ➤ ['Hello', 'Welcome', 'to', 'NLP', '101', 'Let', 's', 'tokenize', 'email', 'example', 'com', 'URLs', 'like', 'https', 'example', 'com']
```

This approach gives you maximal control — you decide exactly what counts as a token (words, numbers, omit punctuation, email, URLs, etc.). Useful if you have domain-specific requirements.

### Which Tokenizer to Use When — Summary

| Tokenization Method          | Use When...                                                                                                                          |
| ---------------------------- | ------------------------------------------------------------------------------------------------------------------------------------ |
| Word-level (NLTK / spaCy)    | Standard tasks in languages with clear spaces (English, etc.), straightforward classification, sentiment analysis, etc.              |
| Sentence + Word (NLTK/spaCy) | When you care about sentence boundaries — summarization, translation per sentence, sentence-level analysis.                          |
| Sub-word (BPE / WordPiece)   | For deep-learning / transformer models, handling rare / unknown words, multilingual or morphologically rich languages.               |
| Character-level              | When dealing with noisy text (typos, misspellings), OCR, or languages without clear boundaries; or for character-level models.       |
| Regex / Custom rule-based    | Domain-specific data (logs, code, medical text), or when you need precise control over tokens (remove emails/URLs/punctuation/etc.). |
