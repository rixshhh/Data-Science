# Word2Vec (CBOW + Skip-gram)

Word2Vec is a neural network model used to learn **word embeddings**—dense vector
representations of words that capture semantic relationships.

### Why Word2Vec?

Before Word2Vec, NLP used Bag-of-Words and TF-IDF, but they had limitations:
- Huge sparse vectors  
- No understanding of meaning  
- Cannot capture context  
- Cannot understand relationships like (king - man + woman = queen)

Word2Vec solves all this using a shallow neural network.

```python
import pandas as pd
import numpy as np
```

```python
import gensim
import os
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

story = []
for filename in os.listdir('data'):
    
    f = open(os.path.join('data',filename), encoding='cp1252', errors='ignore')
    corpus = f.read()
    raw_sent = sent_tokenize(corpus)
    for sent in raw_sent:
        story.append(simple_preprocess(sent))
```

```python
len(story)
```

```python
story
```

```python
model = gensim.models.Word2Vec(
        window=10,
        min_count=2,
        workers=4,
)
```

```python
model.build_vocab(story)
```

```python
model.train(story , total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.wv.most_similar('daenerys')
```

## Comparing CBOW vs SKIP-GRAM


### CBOW (Continuous Bag of Words)
- Predicts the target word from context
- Faster
- Works well for frequent words

### Skip-gram
- Predicts context words from target word
- Slower but more accurate
- Better for rare words
- Most commonly used


## What Word2Vec Learns?

Word2Vec learns:
- Semantic similarity (king ≈ queen, man ≈ woman)
- Analogies (king - man + woman = queen)
- Context relationships
- Meaningful vector space structure

Example:
vec("winterfell") close to vec("north")

vec("cersei") close to vec("lannister")

vec("jon") close to vec("snow")
