# **Lemmatization in NLP (Text Preprocessing)**

Lemmatization is a text preprocessing technique that reduces a word to its **lemma** — the **meaningful base form** of the word.
Unlike stemming, the output of lemmatization **is always a valid dictionary word**.

---

### **Why Lemmatization?**

Lemmatization is used to:

* Understand the real meaning of words
* Normalize words by considering **grammar and context**
* Improve performance in NLP tasks that require semantic understanding
* Reduce sparsity in text datasets

---

### **How Lemmatization Works**

Lemmatization uses:

* **Vocabulary (dictionary lookup)**
* **Part-of-Speech (POS) tagging**
* **Morphological analysis**

Example:

| Word    | POS       | Lemma |
| ------- | --------- | ----- |
| studies | noun      | study |
| better  | adjective | good  |
| running | verb      | run   |
| was     | verb      | be    |

It uses grammar rules to convert words to their correct base form.

---

## **Stemming vs Lemmatization**

| Feature  | Stemming               | Lemmatization              |
| -------- | ---------------------- | -------------------------- |
| Basis    | Rule-based stripping   | Dictionary + POS + grammar |
| Output   | May not be a real word | Always a valid word        |
| Accuracy | Low–Medium             | High                       |
| Speed    | Fast                   | Slower                     |
| Example  | studies → *studi*      | studies → study            |

👉 Lemmatization is preferred for tasks requiring meaning understanding.

---

## **Types of Lemmatizers**

### **1. WordNet Lemmatizer (NLTK)**

Most commonly used English lemmatizer.

Requires POS tag; default is **noun**, so verbs/adjectives need POS input.

### Example:

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

lemmatizer.lemmatize("studies", pos="v")  # → 'study'
lemmatizer.lemmatize("better", pos="a")   # → 'good'
```

### **2. spaCy Lemmatizer**

More advanced:

* Uses large NLP pipelines

* Identifies POS automatically

* Very accurate for modern NLP tasks

Example:

```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("The children were running faster than the others")

for token in doc:
    print(token.text, "→", token.lemma_)
```

## **When to Use Lemmatization?**

Use lemmatization when your task needs **meaning and context**, such as:

* Text classification
* Sentiment analysis
* Chatbots
* Question answering
* Document similarity
* Named entity recognition

---

### **Advantages of Lemmatization**

* Produces meaningful base words
* Context-aware
* Less ambiguity
* Better for semantic NLP tasks
* Works well with deep learning models

---

### **Disadvantages**

* Slower than stemming
* Requires POS tagging
* More computation needed
