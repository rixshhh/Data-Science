#  **What is Bag of Words (BoW) in NLP?**

**Bag of Words (BoW)** is a simple technique to convert text into numerical features based on **word frequency**.
It ignores grammar, word order, and context — it only counts how many times each word appears.

---

##  How BoW Works (Step-by-Step)

### **1. Build Vocabulary**

Collect **all unique words** from the document(s).

Example text:

```
Text 1: "I love India"
Text 2: "India is my country"
```

**Vocabulary:**

```
["I", "love", "India", "is", "my", "country"]
```

### **2. Create vector for each document**

Count how many times each word appears.

| Word   | I | love | India | is | my | country |
| ------ | - | ---- | ----- | -- | -- | ------- |
| Text 1 | 1 | 1    | 1     | 0  | 0  | 0       |
| Text 2 | 0 | 0    | 1     | 1  | 1  | 1       |

These rows are **BoW feature vectors**.

---

## Advantages of BoW

| Advantage             | Explanation                           |
| --------------------- | ------------------------------------- |
| Simple                | Easy to implement                     |
| Effective             | Works well for basic tasks            |
| Good for classical ML | Naive Bayes, SVM, Logistic Regression |

---

## Disadvantages of BoW

| Problem                | Explanation                                |
| ---------------------- | ------------------------------------------ |
| High dimensionality    | Large vocabulary → large vectors           |
| Sparse vectors         | Mostly zeros                               |
| No semantics           | “India” and “country” treated as unrelated |
| Ignores word order     | “dog bites man” = “man bites dog”          |
| Cannot handle synonyms | “big” ≠ “large”                            |

---

## Where BoW is used?

* Text Classification
* Spam Detection
* Sentiment Analysis
* Topic Modeling
* Document Similarity

---

### Comparison: BoW vs TF-IDF vs Word Embeddings

| Feature          | BoW      | TF-IDF              | Embeddings |
| ---------------- | -------- | ------------------- | ---------- |
| Uses frequency   | ✔        | ✔ weighted          | ✔          |
| Captures meaning | ❌        | ❌                   | ✔          |
| Vector size      | Large    | Large               | Small      |
| Best for         | Basic ML | Better classical ML | Deep NLP   |



````python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love India",
    "India is my country"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
````
