# **Stemming in NLP (Text Preprocessing)**

Stemming is a text preprocessing technique used to **reduce words to their root/base form** by **removing prefixes or suffixes**.
The resulting root word **may not be a valid English word**, but it helps in normalizing and grouping similar words together.

---

## **Why Stemming?**

Stemming is used to:

* Reduce vocabulary size
* Group similar words (e.g., “playing”, “played”, “plays” → **play**)
* Improve model performance for search engines, IR systems, and classic ML NLP tasks
* Normalize text before vectorization (Bag-of-Words, TF-IDF)

---

##  **How Stemming Works**

Stemming applies **rule-based stripping** of suffixes.

Examples:

| Word      | Stem    |
| --------- | ------- |
| playing   | play    |
| studies   | studi   |
| waited    | wait    |
| happiness | happi   |
| universal | univers |

Note: Some stems look odd (e.g., *studi*, *univers*) → because stemming is not linguistic.

---

## **Types of Stemmers**

### **1. Porter Stemmer**

* Most popular stemming algorithm
* Introduced by **Martin Porter (1980)**
* Rule-based → removes common suffixes in multiple steps
* Produces aggressive stems

**Example:**

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
ps.stem("studies")   # → 'studi'
```

---

### **2. Snowball Stemmer**

* Also developed by Martin Porter
* Improved version of Porter Stemmer
* Supports multiple languages
* Slightly more accurate and consistent than Porter

**Example:**

```python
from nltk.stem import SnowballStemmer
sb = SnowballStemmer("english")
sb.stem("studies")  # → 'studi'
```

---

### **3. Lancaster Stemmer (Paice-Husk Stemmer)**

* Very **aggressive**
* May produce extremely short stems

**Example:**

| Word    | Lancaster Stem |
| ------- | -------------- |
| maximum | maxim          |
| science | sci            |
| meeting | meet           |

---

## **When to Use Stemming?**

Use stemming when:

* You need **fast preprocessing**
* Exact word meaning **is not important**
* Use cases:

  * Search engines
  * Information retrieval
  * Document classification
  * Text clustering

---

## Stemming vs Lemmatization

| Feature  | Stemming        | Lemmatization      |
| -------- | --------------- | ------------------ |
| Output   | May be non-word | Always proper word |
| Based on | Rules           | Dictionary + POS   |
| Speed    | Fast            | Slower             |
| Accuracy | Medium          | High               |
| Example  | studies → studi | studies → study    |

---

````python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

ps = PorterStemmer()
words = ["playing", "played", "plays", "player"]

for w in words:
    print(w, "→", ps.stem(w))
````


## **Advantages of Stemming**

* Simple & fast
* Reduces dimensionality
* Useful for classic ML algorithms
* Language-independent (rule-based)

---

## **Disadvantages**

* Sometimes inaccurate
* Can distort meaning
* Produces crude word forms
* Not ideal for tasks requiring semantic understanding (e.g., QA, NLP deep learning models)
