# **NLP Text Preprocessing – Notes**

Text preprocessing is the first and most important step in NLP. It transforms raw text into a clean and structured format suitable for machine learning or deep learning models.

---

## 🔹 **1. Introduction**

Text data in the real world is messy. It contains:

* HTML tags
* URLs
* Emojis
* Slang
* Spelling mistakes
* Irrelevant characters

Preprocessing helps in:

* Improving model accuracy
* Reducing noise
* Standardizing text for analysis

---

## 🔹 **2. Lowercasing**

Convert all characters to lowercase.

**Why?**

* Standardizes text
* Reduces vocabulary size
* “Apple” and “apple” become the same token

**Example:**
`"HELLO World"` → `"hello world"`

---

## 🔹 **3. Remove HTML Tags**

Text collected from web pages often contains HTML tags.

**Example:**
`"<p>This is title</p>" → "This is title"`

**Techniques:**

* Regex
* BeautifulSoup

---

## 🔹 **4. Remove URLs**

Texts from social media or websites include links.

**Example:**
`"Check this https://abc.com"` → `"Check this"`

---

## 🔹 **5. Remove Punctuation**

Removes characters like: `! ? , . ; : ' " ( )` etc.

**Why?**

* Punctuation rarely affects sentiment or meaning in ML models
* Reduces noise

**Example:**
`"Hello, world!"` → `"Hello world"`

---

## 🔹 **6. Chat Word Treatment**

Chat and SMS text contains short forms and slang.

**Examples:**

* u → you
* r → are
* btw → by the way
* idk → I don’t know

Use a dictionary to expand chat words.

---

## 🔹 **7. Spelling Correction**

Fixing spelling errors improves token quality.

**Methods:**

* TextBlob
* pyspellchecker
* Custom dictionaries

**Example:**
`"enviroment"` → `"environment"`

---

## 🔹 **8. Removing Stopwords**

Stopwords = common words that do not add meaning.

Examples:

* “the”, “is”, “and”, “in”, “on”

**Why remove?**

* Reduce vocabulary
* Improve model performance

Using libraries:

* NLTK
* SpaCy

---

## 🔹 **9. Handling Emojis**

Emojis carry sentiment.

**Options:**

* Remove
* Convert to text

  * 😀 → “smiling face”
  * ❤️ → “love”

---

## 🔹 **10. Tokenization**

Breaking the text into smaller units.

### Types:

* **Word Tokenization**: "I love NLP" → ["I", "love", "NLP"]
* **Sentence Tokenization**: splits paragraphs into sentences
* **Subword Tokenization**: (BPE used in BERT, GPT)

---

## 🔹 **11. Stemming**

Reducing words to their root form (not always meaningful).

**Examples:**

* “playing” → “play”
* “studies” → “studi”

Libraries:

* Porter Stemmer
* Snowball Stemmer

---

## 🔹 **12. Lemmatization**

Reduces words to meaningful base form.

**Examples:**

* “studies” → “study”
* “better” → “good”

Uses vocabulary + grammar rules.

Libraries:

* WordNet Lemmatizer
* SpaCy
