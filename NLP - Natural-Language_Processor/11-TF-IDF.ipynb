#  **TF-IDF (Term Frequency – Inverse Document Frequency)**

TF-IDF is a text-to-numeric feature extraction method that shows **how important a word is** in a document **relative to the entire corpus**.

👉 It improves Bag of Words by reducing the weight of **common words** and increasing the weight of **rare, informative words**.

---





---

## Simple Example

Documents:

1. `"I love India"`
2. `"India is my country"`

Vocabulary:
`["I", "love", "India", "is", "my", "country"]`

### Step 1: Calculate TF

Example: 
* TF("India") in Doc1 = 1/3

* TF("India") in Doc2 = 1/4

### Step 2: Calculate IDF

"India" appears in **2 docs → df = 2**

[
IDF("India") = \log(2/2) = 0
]

So "India" gets **0 weight** → because it appears everywhere (not informative).

---

### Why TF-IDF is better than BoW?

| Feature             | BoW  | TF-IDF             |
| ------------------- | ---- | ------------------ |
| Raw counts          | ✔    | ❌                  |
| Weighted importance | ❌    | ✔                  |
| Common words impact | High | Low                |
| Ignores stopwords   | No   | Mostly Yes         |
| Useful for ranking  | ❌    | ✔ (Search engines) |

---

### Where TF-IDF is used?

✔ Information Retrieval (Google Search)

✔ Text Classification

✔ Spam Detection

✔ Document Similarity

✔ Recommendation Systems

---

## TF-IDF in Python (sklearn)




```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "I love India",
    "India is my country"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

### Advantages

| Advantage                     | Reason                     |
| ----------------------------- | -------------------------- |
| Removes common word impact    | IDF reduces weight         |
| Improves ML accuracy          | Highlights important terms |
| Simple                        | Easy to compute            |
| Sparse but meaningful vectors | Good for classification    |

---

### Disadvantages

| Disadvantage                             | Why              |
| ---------------------------------------- | ---------------- |
| No semantic meaning                      | “good” ≠ “nice”  |
| Loses word order                         | Bag-based model  |
| High dimensionality                      | Large vocabulary |
| Rare words get too much weight sometimes | Can add noise    |

---

### Summary

**TF-IDF = Better version of Bag of Words**
✔ TF = frequency of word in document

✔ IDF = rarity of word across documents

✔ TF-IDF = importance of word
